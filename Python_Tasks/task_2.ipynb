## Q1. Find the Shortest Word in a Sentence

**Observe:** Check the example input and expected output in the task.

**Understand:** Find the value or result asked for in this question.

**Break:** Split the sentence into words or check the list/string one item at a time.

**Logic:** Split the sentence and keep the shortest word. Using < keeps the first word if there is a tie.

**Code:** The next cell has the Python solution.

**Test:** The code prints the result for the given example and one extra test case.

**Expected results:**
```text
easy
I
```

````python
def solve(s):
    words = s.split()
    if not words: return None
    answer = words[0]
    for w in words[1:]:
        if len(w) < len(answer): answer = w
    return answer
print(solve("Python makes coding easy"))
print(solve("I love Python"))
````

## Q8. Find the Most Frequent Character

**Observe:** Check the example input and expected output in the task.

**Understand:** Find the value or result asked for in this question.

**Break:** Split the sentence into words or check the list/string one item at a time.

**Logic:** Count each character and change the answer only for a higher count, so ties keep the first.

**Code:** The next cell has the Python solution.

**Test:** The code prints the result for the given example and one extra test case.

**Expected results:**
```text
a
a
```

````python
def solve(s):
    chars = [c for c in s if c != " "]
    if not chars: return None
    counts = {}
    for c in chars: counts[c] = counts.get(c, 0) + 1
    answer = chars[0]
    for c in chars:
        if counts[c] > counts[answer]: answer = c
    return answer
print(solve("banana"))
print(solve("aabb"))
````

## Q10. Find All Characters Appearing More Than Once

**Observe:** Check the example input and expected output in the task.

**Understand:** Find the value or result asked for in this question.

**Break:** Split the sentence into words or check the list/string one item at a time.

**Logic:** Track characters already seen and add a repeated character only once.

**Code:** The next cell has the Python solution.

**Test:** The code prints the result for the given example and one extra test case.

**Expected results:**
```text
['r', 'g', 'm']
['l', 'o', 'e']
```

````python
def solve(s):
    seen, result = set(), []
    for c in s:
        if c == " ": continue
        if c in seen and c not in result: result.append(c)
        seen.add(c)
    return result
print(solve("programming"))
print(solve("hello there"))
````

**Expected results:**
```text
{1: 'Odd', 2: 'Even', 3: 'Odd', 4: 'Even', 5: 'Odd'}
```

## Q20. Find the Longest Word That Appears Only Once

**Observe:** Check the example input and expected output in the task.

**Understand:** Find the value or result asked for in this question.

**Break:** Split the sentence into words or check the list/string one item at a time.

**Logic:** Skip words that repeat. If two unique words have the same length, the first one stays as the answer.

**Code:** The next cell has the Python solution.

**Test:** The code prints the result for the given example and one extra test case.

**Expected results:**
```text
coding
green
```

````python
def solve(s):
    words = s.lower().split()
    counts = {}
    for w in words: counts[w] = counts.get(w, 0) + 1
    answer = None
    for w in words:
        if counts[w] == 1 and (answer is None or len(w) > len(answer)):
            answer = w
    return answer
print(solve("Python makes coding useful Python"))
print(solve("red blue red green"))
````
